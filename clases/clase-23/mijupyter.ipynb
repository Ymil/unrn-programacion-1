```python
import matplotlib.pyplot as plt

horas = [0, 1, 2, 3, 4, 5]
temperatura = [18.1, 18.4, 19.0, 20.2, 21.1, 21.5]

plt.plot(horas, temperatura)
plt.title("Temperaturas medidas por hora")
plt.xlabel("Hora")
plt.ylabel("Temp")
plt.show()
```
